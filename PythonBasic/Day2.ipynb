```python
# var 1=45
#function

class Car:
    # attributes  color, model, year
    #methods start stop play

    def __init__(self, color, model, year):
        self.color = color
        self.model = model 
        self.year = year 

    # methods start stop play
    def start(self):
        print("car is starting")

    def stop(self):
        print("car is stoping")

    def play(self, song="Despacito"):
        print(f"Playing {song}")

bmw = Car("red", "X5", 2020) 
```

```python
bmw.start()  # Output: car is starting
bmw.stop()
bmw.play("Despacito")
```

```python
#create a class box and give some attribute and box placing method stack push and pop
# Box Class
class Box:
    def __init__(self, color, length, width, height):
        self.color = color
        self.length = length
        self.width = width
        self.height = height

    def display(self):
        print("Color:", self.color)
        print("Dimensions:", self.length, "x", self.width, "x", self.height)


# Creating an object of Box class
box1 = Box("Red", 10, 5, 8)
box1.display()


# Stack Implementation using List
stack = []

# Push operation
stack.append("Book")
stack.append("Pen")
stack.append("Pencil")

print("Stack after push:", stack)

# Pop operation
item = stack.pop()
print("Popped item:", item)
print("Stack after pop:", stack)
```

```python
#WHILE LOOP USE FOR AND WHILE 


while True:
    user_input = input("Enter a number (or 'q' to quit): ")
    if user_input.lower() == 'q':
        break
    try:
        number = int(user_input)
        print(f"You entered: {number}")
    except ValueError:
        print("Please enter a valid number.")   
```
