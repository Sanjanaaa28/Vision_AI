**What is Convolutional Neural Network?**

A convolutional neural network is a feed-forward neural network that is generally used to analyze visual images by processing data with grid-like topology. It's also known as a ConvNet as a ConvNet. A convolutional neural network is used to detect and classify objects in an image.

```python
import tensorflow as tf
from tensorflow import keras
from matplotlib import pyplot as plt
import numpy as np
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()
```

```python
y_train = y_train.reshape(-1,)
y_train[:5]
```

```python
y_test = y_test.reshape(-1,)
```

```python
classes = ["Airplane", "Automobile","Biird", "Cat", "Deer", "Dog","Frog", "Horse", "Ship", "Trunk"]
```

```python
def plot_sample(x,y,index):
  plt.figure(figsize=(15,2))
  plt.imshow(x[index])
  plt.xlabel(classes[y[index]])
```

```python
plot_sample(x_train,y_train,0)
```

```python
plot_sample(x_train,y_train,1)
```

```python
x_train_Scaled = x_train/255
x_test_Scaled = x_test/255
```

**ANN**

```python
ann = keras.models.Sequential([
    keras.layers.Flatten(input_shape=(32,32,3)),
    keras.layers.Dense(3000,activation="relu"),
    keras.layers.Dense(1000,activation="relu"),
    keras.layers.Dense(10,activation="softmax")
])

ann.compile(
    optimizer="SGD",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

ann.fit(x_train_Scaled,y_train,epochs=10)
```

```python
ann.evaluate(x_test_Scaled,y_test)
```

**CNN**

```python
cnn = keras.models.Sequential([
    keras.layers.Conv2D(filters=32, kernel_size=(3,3), activation="relu",
                        input_shape=(32,32,3)),
    keras.layers.MaxPooling2D((2,2)),

    keras.layers.Flatten(),
    keras.layers.Dense(64,activation="relu"),
    keras.layers.Dense(10,activation="softmax")
])

cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
cnn.fit(x_train_Scaled,y_train,epochs=10)
```

```python
cnn1 = keras.models.Sequential([
    keras.layers.Conv2D(filters=32, kernel_size=(3,3), activation="relu",
                        input_shape=(32,32,3)),
    keras.layers.MaxPooling2D((2,2)),

    keras.layers.Conv2D(filters=64, kernel_size=(3,3), activation="relu"),
    keras.layers.MaxPooling2D((2,2)),

    keras.layers.Flatten(),
    keras.layers.Dense(64,activation="relu"),
    keras.layers.Dense(10,activation="softmax")
])

cnn1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

cnn1.fit(x_train_Scaled, y_train, epochs=10)
```

```python
cnn1.evaluate(x_test_Scaled,y_test)
```

```python
y_pred = cnn.predict(x_test)
y_pred[:5]
```

```python
y_classes = [np.argmax(element) for element in y_pred]
y_classes[:5]
```
