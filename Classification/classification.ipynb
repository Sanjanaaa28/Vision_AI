# Types of Classification

# 1.Binary Classification

when the target variable has exactly two possible outcomes. it comes under binary (0,1)

  - Examples:
     - Email spam detactions
     - Medical Diagnosis (Diseased or Healthy)
     - Application: Logistic Regression, Support vector machines, and Naive bayes

# 2. Multi Class Classification 

when Data need to be categorized into more than two Distinct classes.


Multiple Categories available each data point can still only be assigned to a single or final category.

- Example 1: we have 
    - Animal: Dog , Cat,Horse, Lion
    - still can only assigned to a single outcomes: Cat or Dog or Horse or Lion

- Example 2: we have 
     - Traffic sign: Stop, Go, TurnLeft, Turn Right
     - Still we can assigned 1 outcomes: stop, or go ,or turn left, or turn right

- example 3 : we have 
    - ecom websites ->>>>>> Products categories

- Ques Predict Grade ?
 - A, B, C, D
 - Multi -Class

- Algorithm:
    - Decision Tree, Random Forest and Naive Bayes

# Multi - Label Classification

When the single data point can be assigned multiple labels simultenously.

When we have one record can belong to multiple classes

- Examples:

        - Romance

 - Movie 1 - comedy, Action, Romance as well. ( One record can belong to multiple classes.)       

- Example 2 :

- Example 3:

## Confusion matrix

it is a small table That actual labels with predicted ;abels.

1. True Positive: Predicted positive, actually positive 
2. True Negative: Predicted nagative, actually Negative
3. Flase Positive: Predicted Positive actually negative
4. False Negative: Predicts negative actually positive

- Examples: Email is spam 
    - Actual spam, predicted spam = TP
    - Actual not spam, predicted spam = FP
    - Actual spam, predicted not spam = FN
    - Actual not spam, predicted not spam = TN

## Accuracy

accuracy tell us model correct overall

- Example: if 100 emails, the model gets 90 right, accuracy 90%


- Mislead:
    - accuracy can look high even when the model is poor on the important class.
    - Example: if only 1 out of 100 emails is spam. a model that always says not spam not spam ->>>> Accuracy 99%
    - useless for finding spams!

### Precision:

- When the model predicts POSOTIVE, Right ?
- p = TP/TP+FP

-if Spam filter ->>>> marks 10 emails as spam and 8 really are spam.
- p = 8/10 = 80%

# Recall

- Out of all Real Positive cases, How many did the model find 
- r = TP/TP+FN


- Example:
   - there are 12 spam emails and model catches 9 of them 
   - 9/12 = 75%
