```python
def gini(p,n):
    t = p+n
    p1 = p/t
    p2 = n/t

    g = 1 - (p1**2 + p2**2)
    return g

print("Pure gini: ", gini(10,0))
print("Balanced gini: ", gini(5,5))
print("80-20 gini: ", round(gini(8,2),3))
```
