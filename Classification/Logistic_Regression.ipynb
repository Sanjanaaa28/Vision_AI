```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import(accuracy_score, confusion_matrix, precision_score,classification_report, recall_score)
```

```python
cancer = load_breast_cancer()
```

```python
df = pd.DataFrame(cancer.data, columns = cancer.feature_names)
```

```python
df['Target'] = cancer.target
```

```python
df.head()
```

```python
print("shape:", df.shape)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df['Target'].value_counts())
```

```python
df.head()
```

```python
x = df.drop('Target', axis=1)
y = df['Target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y, test_size = 0.30, random_state = 40
)
```

```python
print(x_train.shape)
```

```python
print(x_test.shape)
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
comparision = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})
```

```python
comparision.head()
```

```python
accuracy = accuracy_score(y_test, y_pred)
```

```python
accuracy # 
```

```python
precision = precision_score(y_test, y_pred)
```

```python
precision
```

```python
confusion_mat = confusion_matrix(y_test, y_pred)
```

```python
confusion_mat
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(confusion_mat, annot=True, cmap = 'Blues',  xticklabels = ["Makignant", 'Benign'], yticklabels=['Makignant','benign'])
plt.xlabel('Predicted')
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
data = {
    "Income": [25000, 30000, 40000,50000,60000,70000,80000],
    "CreditScore": [550,600,620,700,750,800,850],
    "LoanApprove": [0,0,0,0,1,1,1,]
}
```

- Create DF
- X,y
- Split
- train logistic regression
- predict test data
- evaluate confusion matrix, accuracy 
- heatmap

```python
df = pd.DataFrame(data)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.shape)
```

```python
x = df[['Income', 'CreditScore']]
y = df['LoanApprove']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=40
)
```

```python
print(x_train.shape)
```

```python

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print(y_pred)
```

```python
result = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})
result
```

```python
Accuracy= accuracy_score(y_test, y_pred)
```

```python
Accuracy
```

```python
Precision = precision_score(y_test, y_pred)
```

```python
Precision
```

```python
cr = classification_report(y_test,y_pred)
```

```python
Recall = recall_score(y_test, y_pred)
print(Recall)
```

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
print(cm)
```

```python
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',linewidths=0.5,linecolor='black',
            xticklabels=["Not Approved", "Approved"],
            yticklabels=["Not Approved", "Approved"])

plt.title("Loan Approval Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
new_data = [[65000, 780]]

new_data = scaler.transform(new_data)

prediction = model.predict(new_data)

print("Prediction:", prediction)

if prediction[0] == 1:
    print("Loan Approved")
else:
    print("Loan Not Approved")
```
